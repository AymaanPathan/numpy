Generate Random Numbers

```python
import numpy as np
from numpy.random import default_rng
```

```python
randomNum = np.random.random(10)
print(randomNum)
```

Seed is used for reproducibility [to generate a specific set of random number]

```python
range = default_rng(1) ## seed of random number
randomNum = range.random(10)
print(randomNum)
```

range.integers is (start,end,how much do we want) to generate random numbers

```python

specificRandomNum = range.integers(1,101,100)
print(specificRandomNum)
```
