Filtering


```python
import numpy as np
```

```python
salesArr = [0,5,12,34,45,56,0,90]
sales = np.array((salesArr))
print(sales)
```

```python
sales ==0  # gives true / false array
```

```python
sales[(sales <10) | (sales==90)]
```
