```python
prices = np.array([5.99,6.99,22.49,99.99,4.99,49.99])
prices.sort()
threeMostExp = prices[-3:]
mean = prices.mean()
max = prices.max()
min = prices.min()
med = np.median(prices)

# Answers
print("mean" ,mean)
print("threeMostExp" ,threeMostExp)
print("max" , max)
print("min" , min)
print("med" , med)
```
