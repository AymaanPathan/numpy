```python
py_list  =[1,2,3] # Python List
multiply_List = py_list *2
print(multiply_List)
```

```python
numpy_array = np.array([1,2,3]) ## Numpy Array
multiply_NumpyArr = numpy_array *2
print(multiply_NumpyArr)
```

As We See that numpy is correctly doing multiply of arrays

```python
import time
start = time.time()
pyList = [i*2 for i in range(100000)]
print(f"List Operation time {time.time() - start}" )

start =time.time()
npArray = np.arange(100000) *2
print(f"Np Array Operation time {time.time() - start}" )
```

SEQUENCE IN NUMPY

```python
seq = np.arange(1,10 ,2)  # last para is for step [default=1 means it will do one by one]
# in here we have 2 means skip 1 element in every iteration
# range are exclusive outer boundary means it will go till 9 if 10 is end
print(seq)
```

Array Properties

```python
arr1 = [1,2,3]
arr2 = [4,5,6]
array = np.array([arr1,arr2])
print("Shape : ",array.shape)
print("Dimensions : ",array.ndim)
print("Size : ",array.size)
print("Data Type : ",array.dtype)
```

Array Reshaping

```python
arr = np.array([1,2,3,4,5,6])
reshaped = arr.reshape((2,3))
print("Reshaped : \n",reshaped)
```

Flatten Array

```python
flatten = reshaped.flatten() 
print("Flatten : \n",flatten)
```

```python
raveled = reshaped.ravel() # modify the original array
print("reveled : \n " ,raveled)
print(arr)
```
