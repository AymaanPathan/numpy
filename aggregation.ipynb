```python
import numpy as np
```

```python
pricesArr = [5.99,6.99,22.49,99.99,4.99,49.99]
```

```python
prices = np.array((pricesArr))
print(prices)
```

```python
sum  = prices.sum()
mean = prices.mean()
max = prices.max()
min = prices.min()
print("Sum : ",sum)
print("mean : ",mean)
print("max : ",max)
print("min : ",min)
```

2D Array

```python
two_dArray = prices.reshape(3,2)
print(two_dArray)
```

 5.99 + 22.49 + 4.99 = 33.47
 6.99 +99.99 + 49.99 = 156.97

```python
sumOfFirstRowOfAll = two_dArray.sum(axis=0)   
sumOfFirstRowOfAll
```

```python
sumOfFirstColOfAll = two_dArray.sum(axis=1) 
sumOfFirstColOfAll
```
